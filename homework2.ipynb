```python
#Bai 5: giai pt bac 2
import math
a, b, c = map(int,input().split())
if a == 0:
    if b != 0:
        print(f'pt co nghiem {-c/b}')
    if b == 0:
        print(f'vo so nghiem' if c == 0 else f'vo nghiem')
else:
    denta = b*b - 4*a*c
    if denta < 0:
        print("PT vo nghiem")
    elif denta == 0:
        print("PT co nghiem kep:", -b/(2*a))
    else:
        x1 = (-b-math.sqrt(denta))/(2*a)
        x2 = (-b+math.sqrt(denta))/(2*a)
        print(f'PT co 2 nghiem phan biet la {x1} va {x2}')
    
```

```python
#Bai 8: he pttt
a, b, m = map(int,input().split())
c, d, n = map(int,input().split())
det = a*d - b*c
dx = m*d - n*b
dy = a*n - c*m
if det != 0:
    x = dx/det
    y = dy/det
    print(f"Hpt co nghiem duy nhat: x = {x}, y = {y}")
else:
    if dx == 0 and dy == 0:
        print("Hpt co vo so nghiem")
    else:
        print("Hpt vo nghiem")
    
```
