```python
#W6A2
a = list(map(int,input().strip().split()))
s = [0] * len(a)
s[0] = a[0]
for i in range(1, len(a)):
    s[i] = s[i-1] + a[i]
for i in s:
    print(i, end = ' ')
```

```python
#W6A5
a = list(map(int,input().strip().split()))
d = {'positives':0, 'negatives':0, 'zeros':0}
for i in a:
    if i > 0:
        d['positives'] += 1
    elif i < 0:
        d['negatives'] += 1
    else:
        d['zeros'] += 1
print(d)

```

```python
#W6A8
a = list(map(str,input().strip().split()))
d = {}
for i in a:
    if i not in d:
        d[i] = 0
    d[i] += 1
print(d)
```

```python
#W6A9
dic1 = eval(input())
dic2 = eval(input())
dic3 = {}
for k,v in dic1.items():
    dic3[k] = v
for k,v in dic2.items():
    if k not in dic3:
        dic3[k] = v
    else:
        dic3[k] += v
dic3 = dict(sorted(dic3.items()))
print(dic3)
    
```

```python
#W6A12
a = list(map(int,input().strip().split()))
t = [0] * 10**7
maxx = 0
gt = len(a) + 1
for i in range(len(a)):
    t[a[i]] += 1
    if t[a[i]] >= maxx:
        maxx = t[a[i]]
        if a[i] < gt:
            gt = a[i]
print(gt)
```
