```python
#W9A1
def xu_huong(x):
    return x[1] > x[0]
a = list(map(int, input().split()))
b = list(map(int, input().split()))
i = j = 0
c = []
while i < len(a) and j < len(b):
    if xu_huong(a):
        if a[i] < b[j]:
            c.append(a[i])
            i += 1
        else:
            c.append(b[j])
            j += 1
    else:
        if a[i] > b[j]:
            c.append(a[i])
            i += 1
        else:
            c.append(b[j])
            j += 1
while i < len(a):
    c.append(a[i])
    i += 1
while j < len(b):
    c.append(b[j])
    j += 1
for k in c:
    print(k, end = ' ')
            
    
```

```python
#W9A2
A = []
c = []
for i in range(3):
    row = list(map(int, input().split()))
    while len(row) != 5:
        row = list(map(int, input().split()))
    A.append(row)
k = int(input())
for i in A:
    for j in i:
        c.append(j)
c.sort()
print(c[-k])
```

```python
#W9A3
import sys
s = input()
kytu = list(s)
vt = {}
for i in range(len(kytu)):
    vt[int(kytu[i])] = i

for i in range(len(kytu)):
    for j in range(int(kytu[i])):
        if j in vt and vt[j] > i:
            tmp = vt[j]
            kytu[i], kytu[tmp] = kytu[tmp], kytu[i]

            for t in range(len(kytu)):
                if kytu[t] != '0':
                    print("".join(kytu[t:]))
                    break
            sys.exit()

if len(kytu) >= 2:
    kytu[-1], kytu[-2] = kytu[-2], kytu[-1]
    print("".join(kytu))
```

```python
#W9A4
q = int(input())
for i in range(q):
    a = input()
    b = input()
    a = a.lower()
    b = b.lower()
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] == b[j]:
            j += 1
        i += 1
    if j == len(b):
        print("YES")
    else:
        print("NO")
        
```

```python
#W9A8
def is_path_crossing(moves):
    x, y = 0, 0
    a = set()
    a.add((x, y))
    for i in moves:
        if i == 'R':
            x += 1
        elif i == 'L':
            x -= 1
        elif i == 'U':
            y -= 1
        else:
            y += 1

        if (x, y) in a:
            return True
        a.add((x, y))
    return False

s = input()
print(is_path_crossing(s))
        
```
