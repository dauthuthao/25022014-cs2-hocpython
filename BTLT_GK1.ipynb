```python
"""
#Tần số xuất hiện
Nhập vào 1 dãy số đến khi nhập được 1 số âm hoặc >= 10 thì dừng. 
Liệt kê số lần xuất hiện của các số đã nhập theo chiều tăng dần của số.
Không được dùng mảng quá 10 phần tử
"""
t = [0] * 10
n = 0
while n >= 0 and n < 10:
    n = int(input())
    if n >= 0 and n < 10:
        t[n] += 1
for i in range(10):
    if t[i] > 0:
        print(i,':',t[i])
```

```python
"""
#Số Amstrong
Viết hàm kiểm tra một số có phải số Amstrong hay không? 
Hàm trả về kiểu bool, có tham số là một số nguyên x
370: (có 3 chữ số): 3^3 + 7^3 + 0  = 370
"""
def Amst(x):
    s = 0
    x_ = x
    while x != 0:
        s += (x % 10) ** len(str(x_))
        x //= 10
    return s == x_
n = int(input())
print(Amst(n))
```

```python
"""
#Số chính phương
Viết hàm kiểm tra một số có phải số chính phương hay không? 
Hàm trả về kiểu bool, có tham số là một số nguyên x
"""
import math
def scp(x):
    t = math.isqrt(x)
    return t * t == x
n = int(input())
print(scp(n))
```

```python
"""
#Số đối xứng
Viết hàm kiểm tra một số có phải số đối xứng không, 
xử lý trên dữ liệu kiểu số nguyên."""
#cách 1
n = int(input())
n = str(n)
s = ''
for i in range(len(n)-1, -1, -1):
    s += n[i]
if s == n:
    print(True)
else:
    print(False)

#cách 2
n = int(input())
m = 0
while n != 0:
    m = (m * 10) + (n % 10)
    n //= 10
if m == n:
    print(True)
else:
    print(False)
```

```python
"""
#Số kỳ diệu
Viết hàm kiểm tra một số có phải số kỳ diệu hay không,
xử lý trên dữ liệu kiểu số nguyên."""

#TH1: Là số chia hết cho tổng các chữ số của nó.
n = int(input())
n_ = n
s = 0
while n != 0:
    s += n % 10
    n //= 10
if n_ % s == 0:
    print(True)
else:
    print(False)
    
#TH2: Tổng các ước số dương (trừ chính nó) bằng chính nó => số hoàn hảo
import math as m
n = int(input())
def uoc(x):
    s = 1
    for i in range(2, m.isqrt(x) + 1):
        if x % i == 0:
            s += i
            if i != (x/i):
                s += (x/i)
    return s == x
print(uoc(n))
        
```
