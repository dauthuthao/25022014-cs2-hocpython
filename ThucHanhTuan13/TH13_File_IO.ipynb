```python
#W11A1
from pathlib import Path
WORKDIR = Path(".")

path = "W11A1.txt"
def reverse(path: str) -> int:
    file = WORKDIR/path
    with open(file, "r") as f:
        n = int(f.read().strip())
        ans = 0
        while n != 0:
            ans = ans * 10 + (n % 10)
            n //= 10
        return ans
print(reverse(path))
```

```python
#W11A2
from pathlib import Path
WORKDIR = Path(".")

path = "W11A2.txt"

def canWinNim(path: str) -> bool:
    file = WORKDIR/path
    with open(file, "r") as f:
        n = int(f.read().strip())

    if n % 4 == 0:
        return False
    else:
        return True
if canWinNim(path):
    print("Toi thang roi!")
else:
    print("Ban thang roi!")
```

```python
#W11A3
from pathlib import Path
WORKDIR = Path(".")

path = "W11A3.txt"

def housesOfHogwarts(path: str) -> str:
    file = WORKDIR/path
    with open(file, "r", encoding="utf-8") as f:
        lines = f.readlines()
        
    n = int(lines[0].strip())
    for i in range(1, n + 1):
        print(lines[i].strip())
        
housesOfHogwarts(path)
```

```python
#W11A5
#cách đọc input từng dòng => readlines()
from pathlib import Path
WORKDIR = Path(".")

path = "W11A5.txt"

def count(path: str) -> int:
    file = WORKDIR/path
    with open(file, "r") as f:
        lines = f.readlines() 
    s = lines[0].strip()
    c = lines[1].strip()
    ans = s.count(c)
    return ans
print(count(path))
```

```python
#W11A8
from pathlib import Path
WORKDIR = Path(".")

path = "W11A8.txt"

def fibonacci(path: int) -> str:
    in_file = WORKDIR/path
    out_file = WORKDIR/"W11A8_out.txt"
    with open(in_file, "r") as f:
        n = int(f.read().strip()) 
    if n == 0:
        ans = 'a'
    elif n == 1: 
        ans = 'b'
    else:
        s0 = 'a'
        s1 = 'b'
        ans = s0
        for i in range(n - 1):
            ans = s1 + s0
            s0 = s1
            s1 = ans
    with open(out_file, "w") as f:
        f.write(ans)
fibonacci(path)
```
