```python
#W7A16
import math
a = list(map(int,input().split()))
t = [0] * 10**7
s = 0
for i in a:
    t[i] += 1
for i in a:
    if t[i] > 0:
        if i - t[i] > 0:
            s += t[i]
        if i - t[i] < 0:
            s += t[i] - i 
        t[i] = 0
if s == 0:
    print("YES")
else:
    print("NO",s)
```
