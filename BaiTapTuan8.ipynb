```python
#W7A20
def scs(x):
    d = 0
    while x != 0:
        x //= 10
        d += 1
    return d
a = list(map(int,input().split()))
k = int(input())
ans = 0
for i in a:
    if scs(i) == k:
        ans += i
print(ans)
```

```python
#W7A27
def solve(s):
    st = []
    cap_dau = {')': '(', '}': '{', ']': '[', '>': '<'} #dictionary
    for char in s:
        if char in cap_dau:
            if not st: return False
            dau = st.pop()
            if dau != cap_dau[char]: return False
        else:
            st.append(char)
    return len(st) == 0
dau_ngoac = input()
print("YES" if solve(dau_ngoac) else "NO")
```
