```python
#W7A20
def scs(x):
    d = 0
    while x != 0:
        x //= 10
        d += 1
    return d
a = list(map(int,input().split()))
k = int(input())
ans = 0
for i in a:
    if scs(i) == k:
        ans += i
print(ans)
```

```python
#W7A21: tính xem ngày thứ bao nhiêu của năm 2020, nếu hợp lệ
def songay(mm):
    match mm:
        case 1|3|5|7|8|10|12:
            return 31 
        case 2:
            return 29
        case 4|6|9|11:
            return 30
d, m = map(int,input().split())
if (m < 1 or m > 12) or (d < 1 or d > 31):
    print('invalid')
else:
    s = 0
    for i in range(1, m):
        s += songay(i)
    s += d
    print(s)
```

```python
#W7A22
n = int(input())
a = []
for i in range(n):
    s = input()
    s = s.split()
    a.append(s[-1])
a.sort()
d = 1
maxx = 1
vt = 0
for i in range(1, n):
    if a[i] == a[i-1]:
        d += 1
    if maxx < d:
        maxx = d
        vt = i
    if a[i] != a[i-1]:
        d = 1
print(a[vt])
```

```python
# W7A22_cach 2: dùng dict
from collections import Counter
n = int(input())
a = []
for i in range(n):
    s = input()
    s = s.split()
    a.append(s[-1])
a.sort()
count = Counter(a)
ans_name = max(count, key=count.get)
print(ans_name)
```

```python
def solve(s, l, q):
    n = 0
    for j in range(l, l + q):
        n = (n * 10) + (ord(s[j]) - 48)
    return n
st = input()
k = int(input())
a = []
for i in range(len(st) - k + 1):
    if (ord(st[i]) - 48) != 0:
        d = solve(st, i, k)
        a.append(d)
print(max(a))
```

```python
#W7A25
s = input()
k = int(input())
a = []
for i in s:
    a.append(ord(i) - 48)
ans = 0
for i in range(k + 1):
    gt = max(a)
    vt = a.index(gt)
    ans = (ans * 10) + a[vt]
    a.remove(gt)
print(ans)
```
