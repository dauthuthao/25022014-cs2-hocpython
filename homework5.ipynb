```python
#W6A16
n, m = map(int,input().split())
A = []
for i in range(n):
    row = list(map(int,input().strip().split()))
    while len(row) != m:
        row = list(map(int,input(f'Nhập lại hàng {i + 1}:').strip().split()))
    A.append(row)
for i in A:
    for j in i:
        print("{:<4}".format(j),end=' ')  #cách ra để print ra cân đối, đẹp hơn
        #print(j,end = ' ')
    print(end = '\n')
```

```python
#W6A17
n = int(input())
A = [list(map(int,input().strip().split())) for _ in range(n)] # cách khai báo không ktra dc đúng số cột
for i in range(n):
    print(A[i][i], end = ' ')
print(end = '\n')
for i in range(n):
    print(A[i][n-i-1], end = ' ')
```

```python
#W6A18
n, m, k = map(int,input().split())
A = [list(map(int,input().strip().split())) for _ in range(n)]
s1 = s2 = 0
for i in range(n):
    s1 += A[i][k-1]
for i in range(m):
    s2 += A[k-1][i]
print(f'Tổng trên cột {k} là: {s1}')
print(f'Tổng trên hàng {k} là: {s2}')

```
