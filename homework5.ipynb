```python
#W6A16
n, m = map(int,input().split())
A = []
for i in range(n):
    row = list(map(int,input().strip().split()))
    while len(row) != m:
        row = list(map(int,input(f'Nhập lại hàng {i + 1}:').strip().split()))
    A.append(row)
for i in A:
    for j in i:
        print("{:<4}".format(j),end=' ') 
        #print(j,end = ' ')
    print(end = '\n')
```

```python
#W6A18
n, m, k = map(int,input().split())
A = [list(map(int,input().strip().split())) for _ in range(n)]
s1 = s2 = 0
for i in range(n):
    s1 += A[i][k-1]
for i in range(m):
    s2 += A[k-1][i]
print(f'Tổng trên cột {k} là: {s1}')
print(f'Tổng trên hàng {k} là: {s2}')

```

```python
#W6A19 + #W6A22
def maxx(arr):
    ans = arr[0]
    vt = 0
    for i in range(1, len(arr)):
        if arr[i] > ans:
            ans = arr[i]
            vt = i
    return vt
a = list(map(int,input().strip().split()))
print(maxx(a))
    
```

```python
#W6A20
def solve(arr, u):
    for i in range(len(arr)):
        if arr[i] == u:
            return i
    return -1
a = list(map(int,input().strip().split()))
k = int(input())
print(solve(a, k))
```

```python
#W6A21
s = input()
s = s.split()
dict = {}
for i in range(len(s)):
    val = s[i]
    if val not in dict:
        dict[val] = [] #vi tuple k dc thay doi
    dict[val].append(i)
for i in dict:
    dict[i] = tuple(dict[i])
print(dict)
```

```python
#W6A23
def dlt(arr1, x):
    arr2 = []
    for i in arr1:
        if i >= x:
            arr2.append(i)
    return arr2
a = list(map(int,input().strip().split()))
x = int(input())
print(dlt(a, x))
```

```python
#W6A24
#cachs 1: dung ham
def chuyen(ar1):
    ar2 = []
    for i in ar1:
        if i == 0:
            ar2.append(i)
    for i in ar1:
        if i != 0:
            ar2.append(i)
    return ar2
a = list(map(int,input().strip().split()))
a2 = chuyen(a)
for i in a2:
    print(i, end = ' ')

#cach 2: khong dung ham
a = list(map(int,input().strip().split()))
for i in a:
    if i == 0:
        print(i, end = ' ')
for i in a:
    if i != 0:
        print(i, end = ' ')
```

```python
#W6A28
a = list(map(int,input().strip().split()))
tang = giam = bang = True
for i in range(1, len(a)):
    if a[i] != a[i-1]: bang = False
    if a[i] > a[i-1]: giam = False
    if a[i] < a[i-1]: tang = False
if bang:
    print('Day vo danh')
elif giam:
    print('Day giam thuc su')
elif tang:
    print('Day tang thuc su')
else:
    print('Day vo danh')
```
