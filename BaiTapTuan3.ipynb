```python
#W3A11
a, b, c = map(int,input().split())
myl=[]
myl.append(a)
myl.append(b)
myl.append(c)
myl.sort()
x = myl[0]
y = myl[1]
z = myl[2]
if (x + y > z):
    if y == x and y == z:
        print("Tam giac deu")
    elif y == x or y == z:
        print("Tam giac can")
    elif x*x + y*y == z*z:
        print("Tam giac vuong")
    else:
        print("Tam giac thuong")
else:
    print("Khong phai la tam giac")
```

```python
#W3A16 
n = float(input())
if n > 0:
    s1 = int(n) #làm tròn xuống
    s2 = int(n) if n - int(n) == 0 else int(n)+1 #làm tròn lên
    s3 = int(n) if n - int(n) < 0.5 else int(n)+1 #làm tròn đến số gần nhất
else:
    n = n * -1
    s1 = int(n) if n - int(n) == 0 else int(n)+1
    s2 = int(n)
    s3 = int(n)+1 if n - int(n) < 0.5 else int(n)
    s1 *= -1
    s2 *= -1
    s3 *= -1
print(s1,s2,s3)
```
