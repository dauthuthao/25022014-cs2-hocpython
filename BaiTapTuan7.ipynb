```python
#W7A1
def tknp(arr,l,r,x):
    vt = -1
    while l <= r:
        mid = (l + r) // 2
        if arr[mid] == x:
            return mid
        elif arr[mid] > x:
            r = mid - 1
        else:
            l = mid + 1
    return vt
a = list(map(int,input().split()))
k = int(input())
print(tknp(a,0,len(a)-1,k))
```

```python
#W7A2
def count_x(arr,x):
    d = 0
    for i in arr:
        if i == x:
            d += 1
    return d
a = map(int,input().split())
x = int(input())
print(count_x(a,x))
```

```python
#W7A2_cach 3
import bisect as b
a = list(map(int,input().split()))
x = int(input())
a.sort()
vt1 = b.bisect_left(a,x)
vt2 = b.bisect_right(a,x)
ans = vt2 - vt1
print(ans)
```

```python
#W7A3_Bubble Sort
def bubble_sort(arr):
    for i in range(1, len(arr)):
        for j in range(0, i):
            if arr[j] > arr[i]:
                arr[j], arr[i] = arr[i], arr[j]
    return arr
a = list(map(int,input().split()))
bubble_sort(a)
for i in a:
    print(i, end = ' ')
```

```python
#W7A3_Selection sort
def selection_sort(arr):
    for i in range(len(arr)):
        vt_min = i
        for j in range(i + 1, len(arr)):
            if arr[j] < arr[vt_min]:
                vt_min = j
        arr[i], arr[vt_min] = arr[vt_min], arr[i]
    return arr
a = list(map(int,input().split()))
selection_sort(a)
for i in a:
    print(i, end = ' ') 
```

```python
#W7A4
def count_x(arr):
    d = [0] * 10**7
    maxx = -1
    for i in arr:
        d[i] += 1
        if d[i] > maxx:
            maxx = d[i]
    for i in arr:
        if d[i] == maxx:
            gt = i
            break
    return gt, maxx
a =list(map(int,input().split()))
val, ans = count_x(a)
print(f'{val} xuat hien nhieu nhat, som nhat, {ans} lan')
```
