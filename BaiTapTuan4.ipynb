```python
#W4A2
import math
def snt(x):
    if x < 2: return False
    for i in range(2, int(math.sqrt(x))+1):
        if x % i == 0:
            return False
    return True
n = 0
while(n <= 0):
    n = int(input())
if snt(n):
    print(f' {n} la so nguyen to')
else:
    print(f'{n} khong la so nguyen to')
```
