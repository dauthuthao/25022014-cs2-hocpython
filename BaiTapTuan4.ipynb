```python
#W4A2:PrimeNumber
import math
def snt(x):
    if x < 2: return False
    for i in range(2, int(math.sqrt(x))+1):
        if x % i == 0:
            return False
    return True
n = 0
while(n <= 0):
    n = int(input())
if snt(n):
    print(f' {n} la so nguyen to')
else:
    print(f'{n} khong la so nguyen to')
```

```python
#W4A6:SumPrimesInRange

k = int(1e5) + 1
d = [True] * (k + 1)
s = [0] * (k + 1)
def sang():
    d[0] = d[1] = False
    for i in range(2, int(k ** 0.5) + 1):
        if d[i]:
            for j in range(i * i, k + 1, i):
                d[j] = False
def cnt():
    for i in range(2, k + 1):
        if d[i]:
            s[i] = s[i - 1] + i
        else:
            s[i] = s[i - 1]
T = int(input('Nhập số test:'))
sang()
cnt()
for i in range(T):
    a, b = map(int, input().split())
    print(s[b] - s[a - 1])
```

```python
#W4A7:LargestPrimeFactor
def snt(x):
    if x < 2:
        return False
    for i in range(2,int(x**0.5)+1):
        if x % i == 0:
            return False
    return True
n = int(input())
for i in range(n,1,-1):
    if n % i == 0 and snt(i):
        print(i)
        break
```

```python
#W4A8:ReverseAddPalindrome
def rv(x):
    s = 0
    while x != 0:
        s = s*10 + (x % 10)
        x //= 10
    return s
n = int(input())
if n == rv(n):
    print(0)
    print(n)
else:
    m = n + rv(n)
    p = rv(m)
    d = 1
    while m != p:
        n = m
        m = n + rv(n)
        p = rv(m)
        d += 1
    print(d)
    print(p)
```

```python
#W4A9: PerfectSquaresDistinctDigits
def scp(x):
    if int(x ** 0.5) ** 2 == x:
        return True
    else:
        return False
def ktra(d, u):
    while u != 0:
        d[u % 10] += 1
        u //= 10
    for i in range(len(d)):
        if d[i] > 1:
            return False
    return True
kt = False
n = int(input())
for i in range(1,n + 1):
    d = [0] * 11
    if scp(i) and ktra(d, i):
        print(i, end = ' ')
        kt = True
if kt == False:
    print('No number')
```

```python
#W4A10: LongestCollatzUpToN
def collatz(x):
    d = 1
    while x != 1:
        if x % 2 == 0:
            x /= 2
            d += 1
        else:
            x = x * 3 + 1
            d += 1
    return d
n = int(input())
maxx = 0
gt = 0
for i in range(1,n + 1):
    if collatz(i) > maxx:
        maxx = collatz(i)
        gt = i
print(gt,maxx)
# 24 12 6 3 10 5 16 8 4 2 1
# 18 9 28 14 7 22 11 34 17 52 26 13 40 20 10 5 16 8 4 2 1 
```

```python
#W4A13:ClosePairNumber
def t_uoc(x):
    s = 0
    for i in range(1, int(x ** 0.5) + 1):
        if x % i == 0:
            s += i
            if (x/i) != i:
                s += (x/i)
    s -= x
    return s
a, b = map(int,input().split())
if t_uoc(a) == b and t_uoc(b) == a:
    print('true')
else:
    print('false')
```
