```python
#W5A1
def max_of_two(u, v):
    if u > v:
        return u
    else:
        return v
a, b = map(int,input().split())
print(max_of_two(a, b))
```

```python
#W5A2
def swap(u, v):
    t = u
    u = v
    v = t
    return u, v
a, b = map(int,input().split())
ans1, ans2 = swap(a, b)
#print(swap(a, b))
print(ans1, ans2)
```

```python
#W5A3
import math as m
def prime(x):
    if x < 2:
        return False
    for i in range(2, m.isqrt(x) + 1):
        if x % i == 0:
            return False
    return True
n = int(input())
if prime(n):
    print(n, 'la so nguyen to')
else:
    print(n, 'khong la so nguyen to')
```

```python
#W5A4
import math as m
def t_uoc(x):
    s = 0
    for i in range(1, m.isqrt(x) + 1):
        if x % i == 0:
            s += i
            if i != (x/i):
                s += (x/i)
    s -= x
    return s
n = int(input())
if n == t_uoc(n):
    print(n, 'la so hoan hao')
else:
    print(n, 'khong la so hoan hao')
```

```python
#W5A5
a = list(map(int,input().split()))
k = int(input())
kt = False
for i in range(len(a)):
    if a[i] == k:
        print(i+1)
        kt = True
        break
if not kt:
    print(-1)
```

```python
#W5A6
def giaithua(x):
    i = 1
    s = 1
    while i <= x:
        s *= i
        i += 1
    return s
n = int(input())
while n < 0:
    n = int(input('nhập lại: '))
if n == 0:
    print(0)
else:
    print(giaithua(n))    
```

```python
#W5A7
num1, num2 = map(float,input().split())
o = input()
match o:
    case '-':
        print(f'{num1} - {num2} = {round(num1 - num2, 2)}')
    case '+':
        print(f'{num1} + {num2} = {round(num1 + num2, 2)}')
    case '/':
        print(f'{num1} / {num2} = {round(num1 / num2, 2)}')
    case '*':
        print(f'{num1} * {num2} = {round(num1 * num2, 2)}')
```
