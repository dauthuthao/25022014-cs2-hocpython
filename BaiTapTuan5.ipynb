```python
#W5A1
def max_of_two(u, v):
    if u > v:
        return u
    else:
        return v
a, b = map(int,input().split())
print(max_of_two(a, b))
```

```python
#W5A2
def swap(u, v):
    t = u
    u = v
    v = t
    return u, v
a, b = map(int,input().split())
ans1, ans2 = swap(a, b)
print(ans1, ans2)
```

```python
#W5A3
import math as m
def prime(x):
    if x < 2:
        return False
    for i in range(2, m.isqrt(x) + 1):
        if x % i == 0:
            return False
    return True
n = int(input())
if prime(n):
    print(n, 'la so nguyen to')
else:
    print(n, 'khong la so nguyen to')
```

```python
#W5A4
import math as m
def t_uoc(x):
    s = 0
    for i in range(1, m.isqrt(x) + 1):
        if x % i == 0:
            s += i
            if i != (x/i):
                s += (x/i)
    s -= x
    return s
n = int(input())
if n < 2 or n != t_uoc(n):
    print(n, 'khong la so hoan hao')
else:
    print(n, 'la so hoan hao')    
```

```python
#W5A6
def giaithua(x):
    i = 1
    s = 1
    while i <= x:
        s *= i
        i += 1
    return s
n = int(input())
while n < 0:
    n = int(input('nhập lại: '))
if n == 0:
    print(0)
else:
    print(giaithua(n))    
```

```python
#W5A7
num1, num2 = map(float,input().split())
o = input()
match o:
    case '-':
        print(f'{num1} - {num2} = {round(num1 - num2, 2)}')
    case '+':
        print(f'{num1} + {num2} = {round(num1 + num2, 2)}')
    case '/':
        print(f'{num1} / {num2} = {round(num1 / num2, 2)}')
    case '*':
        print(f'{num1} * {num2} = {round(num1 * num2, 2)}')
```

```python
#W5A8
#xor 2 bit khác nhau kqua = 1
def hamming(u, v):
    p = u ^ v
    ans = 0
    while p != 0:
        ans += p % 2
        p //= 2
    return ans
x, y = map(int, input().split())
#print(x ^ y)
print(hamming(x, y))
```

```python
#W5A10 => cách dùng mảng đánh dấu
s1_s2 = [None] * 256
s2_s1 = [None] * 256
s1, s2 = map(str, input().split())
for i in range(len(s1)):
    chr1 = s1[i]
    chr2 = s2[i]
    if s1_s2[ord(chr1)] is None:
        s1_s2[ord(chr1)] = chr2
    elif s1_s2[ord(chr1)] != chr2:
        print('False')
        break
    if s2_s1[ord(chr2)] is None:
        s2_s1[ord(chr2)] = chr1
    elif s2_s1[ord(chr2)] != chr1:
        print('False')
        break
else:
    print('True')
```
