```python
#Bai4
try:
    h = int(input("Nhập chiều cao tam giác: "))
    for i in range(h):
        if i == h - 1:
            print('*' * (2 * h - 1))
        else:
            print(' ' * (h - i - 1), end='')
            if i == 0:
                print('*')
            else:
                print('*' + ' ' * (2 * i - 1) + '*')
except ValueError:
    print("Vui lòng nhập số nguyên dương!")
```

```python
#Bai14
n = 0
maxx = -1e10
minn = 1e10
while n != -1:
    n = int(input())
    if n > maxx: maxx = n
    if n < minn: minn = n
# ??? -1 có được tính vào không, nếu -1 chỉ để dừng nhập không cần 
# cập nhật maxx & minn, break ngay khi n = -1
# nhưng nếu -1 vẫn được tính vào dãy, vẫn phải cập nhật maxx & minn
print("so lon nhat la: ",maxx)
print("so nho nhat la: ",minn)
```

```python
#Bai18: kiem tra xem n co dang 3^k khong
n = int(input())
s = n
d = 0
while n % 3 == 0:
    d += 1
    n //= 3
if n == 1:
    print(f'{s} co dang 3^k, {s} = 3^{d}')
else:
    print(f'{s} khong co dang 3^k')
```

```python
#Bai19: tim so trong day Fibonacci lon nhat nhung khong vuot qua A
t1 = t2 = 1
t = 0
a = int(input())
if a == 1:
    print(1)
else:
    while t2 <= a:
        t  = t1
        t1 = t2
        t2 = t + t1
    print(t2 - t)
```

```python
#Bai20
def snt(n):
    if n < 2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n % i == 0:
            return False
    return True
n = int(input())
s = n
d = 0
while n != 0:
    if snt(n % 10): d += 1
    n //= 10
print(f'{s} co {d} chu so la so nguyen to')
```
