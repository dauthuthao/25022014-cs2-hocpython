```python
#kiem tra ngay thang hop le
#cach 1: dung if else
#cach 2: dung match...case
dd, mm, yy = map(int,input().split('/'))
if yy >= 1900:
    match mm:
    #thang 31 ngay
        case 1|3|5|7|8|10|12:
            print("Valid" if 1 <= dd <= 31 else "Invalid")
    #thang 30 ngay
        case 4|6|9|11:
            print("Valid" if 1 <= dd <= 30 else "Invalid")
    #thang 2
        case 2:
            check = (yy % 400 == 0) or (yy % 4 == 0 and yy % 100 !=0)
            max_day = 29 if check else 28
            print("Valid" if 1 <= dd <= max_day else "Invalid")
            
        case _: print("Invalid")
else:
    print("Invalid")
```
