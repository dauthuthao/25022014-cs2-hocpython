```python
#Bai8
name = input('Ten chu ho: ')
cs_truoc = int(input('Chi so thang truoc: '))
cs_sau = int(input('Chi so thang sau: '))
t = cs_sau - cs_truoc
s = 0
if t <= 50:
    s = t * 1984
if t > 50 and t <= 100:
    s = 50*1984 + (t-50)* 2050
if t > 100 and t <= 200:
    s = 50*1984 + 50*2050 + (t-100)*2380
if t > 200 and t <= 300:
    s = 50*1984 + 50*2050 + 100*2380 + (t-200)*2998
if t > 300 and t <= 400:
    s = 50*1984 + 50*2050 + 100*2380 + 100*2998 + (t-300)*3350
if t > 400:
    s = 50*1984 + 50*2050 + 100*2380 + 100*2998 + 100*3350 + (t-400)*3460
    
s = s + (s*8)/100
if s-int(s) >= 0.5:
    s = int(s) + 1
print('Ho va ten: ',name)
print('Tien phai tra la: ',int(s))
```
