```python
def kytu(x):
    if x == ')': return '('
    elif x == ']': return '['
    else: return '{'

s = input()
ch = [')', ']', '}']
a = []

for i in s:
    if i in ch:
        if len(a) == 0:
            print("Invalid")
            exit()
        if a[-1] != kytu(i):
            print("Invalid")
            exit()
        a.pop()
    else:
        a.append(i)

if len(a) == 0:
    print("Valid")
else:
    print("Invalid")
```

```python
# so sánh tổng trái - phải
n = int(input())
n = str(n)
vt = len(n)//2 - 1
tong = t1 = t2 = 0

for i in range(len(n)):
    tong += (ord(n[i]) - 48)

for i in range(vt + 1):
    t1 += (ord(n[i]) - 48)

if len(n) % 2 == 0:
    t2 = tong - t1
else:
    t2 = tong - t1 - (ord(n[vt + 1]) - 48)

if t1 == t2:
    print("True")
else:
    print("False")
```

```python
#dãy con tăng nghiêm ngặt, dài nhất
a = eval(input())
L = [1] * len(a)

for i in range(1, len(a)):
    for j in range(i):
        if a[j] < a[i] and L[i] < L[j] + 1:
            L[i] = L[j] + 1

print(max(L))
```

```python
# đếm số phầm sử nhỏ hơn bên phải
a = eval(input())
ans = []
t = 0

for i in range(len(a) - 1):
    for j in range(i + 1, len(a)):
        if a[j] < a[i]:
            t += 1
    ans.append(t)
    t = 0

ans.append(0)
print(ans)
```
