```python
#Bai31
#tinh so fibo thu n
def fibo(t):
    if t == 1 or t == 2:
        return 1
    else:
        t1 = t2 = 1
        t3 = 0
        for i in range(3,t+1):
            t3 = t1
            t1 = t2
            t2 = t3 + t1
        return t2
n = int(input())
print(fibo(n))
```

```python
#Bai32_Bai2: kiem tra ngay thu may trong nam 2000
def songay(mm):
    match mm:
        case 2: 
            return 29
        case 4|6|9|11:
            return 30
        case 1|3|5|7|8|10|12:
            return 31
dd, mm = map(int, input().split('/'))
if (mm < 1 or mm > 12) or (dd < 1 or dd > songay(mm)):
    print('Khong hop le!')
else:
    s = 0
    for i in range(1, mm):
        s += songay(i)
    s += dd
    print(f'{dd}/{mm} la ngay thu {s} cua nam 2000')
```

```python
#Bai35
def cnt(x,n):
    d1 = d2 = 1
    ans = 0
    for i in range(1,n+1):
        d1 *= x
        d2 *= i + 1
        ans += (1 + d1)/d2
    return ans
x, n = map(int,input('Nhap x va n:').split())
if n == 0:
    print('Tong S = ',2)
else:
    print(cnt(x,n) + 2)
```

```python
#Bai46_Cách 3
#2 con trỏ, ktra từ đầu đến giữa - từ cuối lên giữa
s = input('Nhập chuỗi kí tự:')
n = len(s)
kt = True
vt = 0
if n % 2:
    vt = n//2 - 1
else: 
    vt = n//2
for i in range(vt + 1):
    if s[i] != s[n - 1 - i]:
        kt = False
        break
if kt:
    print(s,'là xâu đối xứng')
else:
    print(s,'không là xâu đối xứng')
```
