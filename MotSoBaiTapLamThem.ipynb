```python
#Bai19
import math
def snt(t):
    if t < 2:
        return False
    for i in range(2, int(math.sqrt(t)+1)):
        if t % i == 0:
            return False
    return True
n = int(input())
if snt(n):
    print(f'{n} la so nguyen to')
else:
    print(f'{n} khong la so nguyen to')
```

```python
#Bai20
import math
def snt(t):
    if t < 2:
        return False
    for i in range(2, int(math.sqrt(t)+1)):
        if t % i == 0:
            return False
    return True
n = int(input())
for i in range(2,n+1):
    if snt(i):
        print(i)
```
