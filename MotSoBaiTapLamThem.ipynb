```python
#Bai19
import math
def snt(t):
    if t < 2:
        return False
    for i in range(2, int(math.sqrt(t)+1)):
        if t % i == 0:
            return False
    return True
n = int(input())
if snt(n):
    print(f'{n} la so nguyen to')
else:
    print(f'{n} khong la so nguyen to')
```

```python
#Bai31
#tinh so fibo thu n
def fibo(t):
    if t == 1 or t == 2:
        return 1
    else:
        t1 = t2 = 1
        t3 = 0
        for i in range(3,t+1):
            t3 = t1
            t1 = t2
            t2 = t3 + t1
        return t2
n = int(input())
print(fibo(n))
```

```python
#Bai31: ktra 1 so co la so Fibonacci khong
n = int(input())
t1 = t2 = 1
t = 0
while t2 < n:
    t = t1
    t1 = t2
    t2 = t + t1
if n == t2:
    print(n,'la so Fibonacci')
else:
    print(n,'khong la so Fibonacci')
```

```python
#Bai35
def cnt(x,n):
    d1 = d2 = 1
    ans = 0
    for i in range(1,n+1):
        d1 *= x
        d2 *= i + 1
        ans += (1 + d1)/d2
    return ans
x, n = map(int,input('Nhap x va n:').split())
if n == 0:
    print('Tong S = ',2)
else:
    print(cnt(x,n) + 2)
```

```python
#Bai36: giai hpt
a11, a12, c1 = map(int,input('Nhap vao a11, a12, c1: ').split())
a21, a22, c2 = map(int,input('Nhap vao a21, a22, c2: ').split())
det = a11 * a22 - a21 * a12
if det != 0:
    x1 = (c1 * a22 - c2 * a12)/det
    x2 = (a11 * c2 - a21 * c1)/det
    print(f'He pt co nghiem duy nhat: x1 = {x1}, x2 = {x2}')
else:
    if a11/a21 == a12/a22 == c1/c2:
        print('He pt co vo so nghiem')
    else:
        print('He pt vo nghiem')
```
