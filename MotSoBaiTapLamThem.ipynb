```python
#Bai19
import math
def snt(t):
    if t < 2:
        return False
    for i in range(2, int(math.sqrt(t)+1)):
        if t % i == 0:
            return False
    return True
n = int(input())
if snt(n):
    print(f'{n} la so nguyen to')
else:
    print(f'{n} khong la so nguyen to')
```

```python
#Bai26 cach 2
def count(b,m):
    if m == 0:
        return 1
    elif m == 1:
        return b
    else:
        return b * count(b,m-1)
a, n = map(int,input().split())
print(count(a,n))
```

```python
#Bai30
print('0. Không làm gì')
print('1. Tính diện tích hình vuông')
print('2. Tính diện tích hình tròn')
print('3. Tính diện tích tam giác')
print('4. Tính diện tích hình chữ nhật')
n = int(input('Chọn lệnh'))
if n == 0:
    print('0. Không làm gì')
    print('1. Tính diện tích hình vuông')
    print('2. Tính diện tích hình tròn')
    print('3. Tính diện tích tam giác')
    print('4. Tính diện tích hình chữ nhật')
if n == 1:
    print(f'Diện tích hình vuông: S = a * a')
if n == 2:
    print(f'Diện tích hình tròn: S = r * r * 3.14')
if n == 3:
    print(f'Diện tích tam giác: S = math.sqrt(p*(p-a)*(p-b)*(p-c)) với p = (a + b + c)/2')
if n == 4:
    print(f'Diện tích hình chữ nhật: S = d * r')
    
```

```python
#Bai31
#tinh so fibo thu n
def fibo(t):
    if t == 1 or t == 2:
        return 1
    else:
        t1 = t2 = 1
        t3 = 0
        for i in range(3,t+1):
            t3 = t1
            t1 = t2
            t2 = t3 + t1
        return t2
n = int(input())
print(fibo(n))
```
