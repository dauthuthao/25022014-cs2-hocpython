```python
def tcs(x):
    s = 0
    while x != 0:
        s += x % 10
        x //= 10
    return s

n = int(input().strip())
a = list(map(int, input().split()))
A = []
a.sort()
for i in a:
    A.append([i, tcs(i)])

A.sort(key=lambda x: x[1])
for i in A:
    print(i[0], end=' ')
```

```python
import math

def closest_value(a, x):
    minn = abs(a[0] - x)
    for i in a:
        if abs(i - x) < minn:
            minn = abs(i - x)

    ans = max(a)
    for i in a:
        if abs(i - x) == minn and i < ans:
            ans = i
    return ans

n = int(input().strip())
a = list(map(int, input().split()))
x = int(input().strip())
print(closest_value(a, x))
```

```python
def top_k_desc(a):
    for i in range(len(a) - 1):
        for j in range(i + 1, len(a)):
            if a[j] > a[i]:
                tmp = a[j]
                a[j] = a[i]
                a[i] = tmp
    return a

n, k = map(int, input().split())
a = list(map(int, input().split()))
b = top_k_desc(a)
for i in range(k):
    print(b[i], end=' ')
```

```python
def count_digits(n):
    if n < 10:
        return 1
    return 1 + count_digits(n // 10)

n = int(input().strip())
print(count_digits(n))
```

```python
def is_pal_ign_spaces(s):
    ans = ''
    for i in range(len(s) - 1, -1, -1):
        ans += s[i]
    return ans == s

s = input().rstrip("\n")
print(is_pal_ign_spaces(s))
```

```python
def sum_odds(a, i=0):
    if i == len(a):
        return 0
    if a[i] % 2 != 0:
        return a[i] + sum_odds(a, i + 1)
    return sum_odds(a, i + 1)

n = int(input().strip())
a = list(map(int, input().split())) if n > 0 else []
print(sum_odds(a))
```

```python
n = int(input().strip())
d = dict()
for i in range(n):
    s = input()
    if s[0] not in d:
        d[s[0]] = []
    d[s[0]].append(s)

for k in sorted(d):
    print(k + ':' + ','.join(d[k]))
```

```python
n = int(input())
d = dict()
for i in range(n):
    c, s = map(str, input().split())
    if c not in d:
        d[c] = 0
    d[c] += int(s)

for k, v in d.items():
    print(k, v)
```
